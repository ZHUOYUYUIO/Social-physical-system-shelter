```python

import geopandas as gpd
import matplotlib.pyplot as plt

# 加载第一个GeoJSON文件
gdf1 = gpd.read_file('transformed_buildings.geojson')  # 替换为你的第一个GeoJSON文件路径

# 加载第二个GeoJSON文件
gdf2 = gpd.read_file('transformed_boundary.geojson')  # 替换为你的第二个GeoJSON文件路径

# 创建图形和坐标轴
fig, ax = plt.subplots(figsize=(10, 10))

# 绘制第一个GeoJSON
gdf1.plot(ax=ax, color='blue', alpha=0.5, label='Layer 1')

# 绘制第二个GeoJSON
gdf2.plot(ax=ax, color='red', alpha=0.5, label='Layer 2')

# 添加图例和标题
ax.legend()
plt.title('Overlay of Two GeoJSON Files')
plt.show()
```

```python
# 获取总边界
bounds = gdf2.total_bounds
print("社区边界坐标 (minx, miny, maxx, maxy):", bounds)
```

```python
bounds = gdf1.total_bounds
print("建筑物边界坐标 (minx, miny, maxx, maxy):", bounds)
```
